```python
import numpy as np
import tensorflow as tf

from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, GRU, Dense
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
vocab_size = 10000

(X_train, y_train), (X_test, y_test) = imdb.load_data(
    num_words=vocab_size
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
print("First review:", X_train[0])
print("First label:", y_train[0])
```

```python
max_length = 200

X_train = pad_sequences(
    X_train,
    maxlen=max_length,
    padding="post",
    truncating="post"
)

X_test = pad_sequences(
    X_test,
    maxlen=max_length,
    padding="post",
    truncating="post"
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

```python
model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=128),
    LSTM(64),
    Dense(1, activation="sigmoid")
])

model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
history = model.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.2
)
```

```python
test_loss, test_accuracy = model.evaluate(X_test, y_test)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
```

```python
predictions = model.predict(X_test)

predicted_labels = (predictions > 0.5).astype(int)

print("First 10 predictions:")
print(predicted_labels[:10].flatten())

print("Actual labels:")
print(y_test[:10])
```

```python
model.save("lstm_sentiment_model.keras")
```

```python
gru_model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=128),
    GRU(64),
    Dense(1, activation="sigmoid")
])

gru_model.summary()
```

```python
gru_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
gru_history = gru_model.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.2
)
```

```python
gru_loss, gru_accuracy = gru_model.evaluate(X_test, y_test)

print("GRU Test Loss:", gru_loss)
print("GRU Test Accuracy:", gru_accuracy)
```

```python
print("LSTM Test Accuracy:", test_accuracy)
print("GRU Test Accuracy:", gru_accuracy)
```

```python
if gru_accuracy > test_accuracy:
    print("GRU performed better than LSTM.")
elif test_accuracy > gru_accuracy:
    print("LSTM performed better than GRU.")
else:
    print("Both models performed equally.")
```

```python
def predict_sentiment(review, model):
    words = review.lower().split()

    word_index = imdb.get_word_index()

    encoded_review = [
        word_index.get(word, 2) + 3
        for word in words
    ]

    encoded_review = pad_sequences(
        [encoded_review],
        maxlen=200,
        padding="post",
        truncating="post"
    )

    prediction = model.predict(encoded_review, verbose=0)[0][0]

    if prediction >= 0.5:
        return "Positive"
    else:
        return "Negative"
```

```python
review = "This movie was amazing and I really enjoyed it."

result = predict_sentiment(review, gru_model)

print("Review:", review)
print("Sentiment:", result)
```

```python
model.save("lstm_sentiment_model.keras")
gru_model.save("gru_sentiment_model.keras")

print("Both models saved successfully.")
```
